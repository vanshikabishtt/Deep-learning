```python
questions = [i['question'] for i in data]
answers = [i['answer'] for i in data]

tokenizer = Tokenizer()
tokenizer.fit_on_texts(questions + answers)

vocab_size = len(tokenizer.word_index) + 1

X = tokenizer.texts_to_sequences(questions)
y = tokenizer.texts_to_sequences(answers)

max_len = 12
X = pad_sequences(X, maxlen=max_len, padding='post')
y = pad_sequences(y, maxlen=max_len, padding='post')
```

```python
model = Sequential()
model.add(Embedding(vocab_size, 64, input_length=max_len))
model.add(GRU(128, return_sequences=True))
model.add(Dense(vocab_size, activation='softmax'))

model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

y = np.expand_dims(y, -1)

model.fit(X, y, epochs=300, verbose=1)
```

```python
def chatbot_response(user_input):
    user_input = user_input.lower()

    # Greetings
    if user_input in ["hi", "hello", "hey"]:
        return "Hello! 👋 I can recommend movies by genre. Try 'action movie' or 'sci fi movie'."

    # Thanks
    if "thank" in user_input:
        return "You're welcome 😊 Have a great day and enjoy your movies!"

    # Goodbye
    if user_input in ["bye", "goodbye"]:
        return "Goodbye! 🎬 Have a great day!"

    # Movie recommendation
    for item in data:
        if item['question'] in user_input:
            sorted_movies = sorted(item['ratings'].items(), key=lambda x: x[1], reverse=True)
            return "Top rated: " + " | ".join([f"{m} ({r})" for m, r in sorted_movies])

    return "Sorry, I don't understand. Try genres like action, sci-fi, comedy."
```
